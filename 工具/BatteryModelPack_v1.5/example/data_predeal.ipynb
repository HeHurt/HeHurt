```python
import numpy as np
import pandas as pd

for i in ['1']:
    soh = np.loadtxt('data/Age/%sP.dat' % i)
    cycle = np.arange(1, len(soh) + 1)
    data = pd.DataFrame({'0': cycle, 'soh': soh})
    data.to_csv('data/Age/%sP.dat' % i, index=False, header=False, sep='\t')
```

```python
import matplotlib.pyplot as plt
for i in ['15', '25', '35', '45']:
    data = pd.read_csv(r'G:\Code\BatteryModel\notebooks\314\diff_temperature\data\%s.dat' % i, sep=',')
    # cycle = data['圈数'].to_numpy()
    soh = data['放电容量'].to_numpy()
    soh = soh / soh[0]
    if i == '15':
        soh[soh < 0.96] = soh[soh < 0.96] + 0.0055
    data = pd.DataFrame({'0': np.arange(1, len(soh) + 1, 1), '1': soh})
    data.to_csv('data/Age_diffT/%s.dat' % i, index=False, header=False, sep='\t')
    plt.plot(soh)
```
