## 4. 快速 EIS 基线

单温度、单 SOC 的 Nyquist 曲线用于确认 EIS 方程和双层电容参数可用。


```python
quick_df = quick_impedance_df.loc[quick_impedance_df["section"] == "quick"]
if quick_df.empty:
    print("快速 EIS 未启用。")
else:
    fig, ax = plt.subplots(figsize=(6.5, 5))
    for label, frame in quick_df.groupby("label", sort=False):
        ax.plot(frame["z_real_ohm"], frame["minus_z_im_ohm"], "-o", ms=3, label=label)
    ax.set_xlabel("Z' / Ω")
    ax.set_ylabel("-Z'' / Ω")
    ax.set_aspect("equal")
    ax.set_title("快速 EIS 基线")
    ax.grid(alpha=0.3)
    ax.legend()
    plt.tight_layout()
    plt.show()
    display(summarize_quick_components(quick_df))
```

## 5. SOC 扫描

固定温度，只改变初始 SOC，用于检查阻抗随 SOC 的趋势。


```python
soc_df = quick_impedance_df.loc[quick_impedance_df["section"] == "soc_sweep"]
if soc_df.empty:
    print("SOC 扫描未启用。")
else:
    fig, ax = plt.subplots(figsize=(7, 5))
    for label, frame in soc_df.groupby("label", sort=False):
        ax.plot(frame["z_real_ohm"], frame["minus_z_im_ohm"], "-o", ms=3, label=label)
    ax.set_xlabel("Z' / Ω")
    ax.set_ylabel("-Z'' / Ω")
    ax.set_aspect("equal")
    ax.set_title(f"SOC 扫描 @ {spec.base_temperature_c:g}°C")
    ax.grid(alpha=0.3)
    ax.legend()
    plt.tight_layout()
    plt.show()
```

## 7. Lifecycle-EIS 联合研究

在循环衰减过程中抽取 SOH 检查点，并在每个检查点测 EIS。


```python
if lifecycle_result is None:
    print("Lifecycle-EIS 未启用。")
else:
    lifecycle_soh_df = lifecycle_result["soh_df"]
    lifecycle_checkpoints_df = lifecycle_result["checkpoints_df"]
    lifecycle_impedance_df = lifecycle_result["impedance_df"]
    lifecycle_components_df = lifecycle_result["components_df"]
    state_label = lifecycle_result["measurement_state_label"]

    fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
    axes[0].plot(lifecycle_soh_df["cycle"], lifecycle_soh_df["soh_pct"], "-o", ms=3, color="#1f77b4")
    for _, row in lifecycle_checkpoints_df.iterrows():
        cycle_number = int(row["cycle"])
        axes[0].scatter(cycle_number, row["soh_pct"], s=60, color="#d62728", zorder=3)
        axes[0].annotate(f"{row['soh_pct']:.1f}%", (cycle_number, row["soh_pct"]), textcoords="offset points", xytext=(0, 8), ha="center", fontsize=8)
        nyquist_frame = lifecycle_impedance_df.loc[lifecycle_impedance_df["label"] == row["label"]]
        axes[1].plot(nyquist_frame["z_real_ohm"], nyquist_frame["minus_z_im_ohm"], "-o", ms=3, label=row["label"])
    axes[0].set_xlabel("Cycle number")
    axes[0].set_ylabel("SOH (%)")
    axes[0].set_title("Lifecycle fade")
    axes[0].grid(alpha=0.3)
    axes[1].set_xlabel("Z' / Ω")
    axes[1].set_ylabel("-Z'' / Ω")
    axes[1].set_title(f"Lifecycle Nyquist ({state_label})")
    axes[1].grid(alpha=0.3)
    axes[1].legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    display(lifecycle_checkpoints_df)
    display(lifecycle_components_df)
    print(f"aging_power_w = {lifecycle_result['aging_power_w']:.1f}")
    print(f"EIS measurement state = {state_label}")
```

## 8. 阻抗指标拆解

这里是工程代理指标：Rs / Rct proxy / diffusion proxy 来自 Nyquist 几何特征，不等价于严格物理解耦。


```python
if lifecycle_result is None or lifecycle_result["components_df"].empty:
    print("没有 lifecycle component 表可拆解。")
else:
    component_plot_df = lifecycle_result["components_df"].sort_values("soh_pct", ascending=False).reset_index(drop=True)
    fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
    x = np.arange(len(component_plot_df))
    axes[0].bar(x, component_plot_df["r_ohm_ohm"], label="R_ohm")
    axes[0].bar(x, component_plot_df["r_ct_proxy_ohm"], bottom=component_plot_df["r_ohm_ohm"], label="R_ct proxy")
    axes[0].bar(x, component_plot_df["r_diffusion_proxy_ohm"], bottom=component_plot_df["r_ohm_ohm"] + component_plot_df["r_ct_proxy_ohm"], label="R_diffusion proxy")
    axes[0].set_xticks(x)
    axes[0].set_xticklabels([f"{soh:.1f}%" for soh in component_plot_df["soh_pct"]])
    axes[0].set_xlabel("SOH")
    axes[0].set_ylabel("Resistance / Ω")
    axes[0].set_title("Impedance components vs SOH")
    axes[0].legend()
    axes[0].grid(axis="y", alpha=0.3)
    axes[1].plot(component_plot_df["soh_pct"], component_plot_df["z_magnitude_at_1hz_ohm"], "-o", color="#ff7f0e", label="|Z| @ 1 Hz")
    axes[1].plot(component_plot_df["soh_pct"], component_plot_df["r_polarization_ohm"], "-s", color="#2ca02c", label="R_polarization")
    axes[1].set_xlabel("SOH (%)")
    axes[1].set_ylabel("Impedance / Ω")
    axes[1].set_title("Key impedance indicators vs SOH")
    axes[1].invert_xaxis()
    axes[1].grid(alpha=0.3)
    axes[1].legend()
    plt.tight_layout()
    plt.show()

    display(frequency_slice_df)
    print("If the -Im(Z) peak falls on the scan boundary, R_ct proxy is set to 0 and the non-ohmic rise is assigned to diffusion/tail.")
```
